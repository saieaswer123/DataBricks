1. WE HAVE TO FIND THE HIGHEST SALARY BASED ON EACH DEPARTMENT NAME
2. WE HAVE TO FIND THE EMPLOYEE WHO IS GETTING HIGHEST SALARY BASED ON EACH DEPARTMENT NAME
3. WE HAVE TO FIND THE lowest SALARY BASED ON EACH DEPARTMENT NAME
4. WE HAVE TO FIND THE EMPLOYEE WHO IS GETTING lowest SALARY BASED ON EACH DEPARTMENT NAME
SOLVE USING PYSPARK AND SPARK SQL

```python
emp_data=[('Manish' , 1 , 75000),
('Raghav' , 1 , 85000 ),
('surya' , 1 , 80000 ),
('virat' , 2 , 70000),
('rohit' , 2 , 75000),
('jadeja' , 3 , 85000),
('anil' , 3 , 55000),
('sachin' , 3 , 55000),  
('zahir', 4, 60000),
('bumrah' , 4 , 65000) ]
schema= ["emp_name" ,"dept_id" ,"salary"]

emp_df=spark.createDataFrame(emp_data,schema)
emp_df.show()
```

```python
dept_data = [(1, 'DATA ENGINEER'),(2, 'SALES'),(3, 'SOFTWARE'),(4, 'HR')]
schema1=['dept_id','dept_name']

dept_df=spark.createDataFrame(dept_data,schema1)
dept_df.show()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
res=emp_df.join(dept_df,emp_df.dept_id==dept_df.dept_id,'inner')
```

```python
res.show()
```

```python
window_spe=Window.partitionBy('dept_name').orderBy(desc('salary'))

res.withColumn('rank',dense_rank().over(window_spe)).filter('rank==1').show()
```

```python
res.groupBy("dept_name").max('salary').show()
```

```python
window=Window.partitionBy('dept_name').orderBy('salary')

res.withColumn('rank',dense_rank().over(window)).filter('rank = 1').show()
```

```python
res.groupBy("dept_name").min('salary').show()
```

```python
res.createOrReplaceTempView("res")
```

```python
%sql
select * from res
```

```python
%sql
select emp_name,salary from (select emp_name,salary,row_number() over(partition by dept_name order By salary desc) as rank from res) where rank=1
```

```python
%sql
select emp_name,salary,dept_name from res group by(dept_name) having max(salary)
```

```python
%sql
select emp_name,salary from(select emp_name,salary,row_number() over(partition By dept_name order by salary) as rank from res) where rank=1
```
