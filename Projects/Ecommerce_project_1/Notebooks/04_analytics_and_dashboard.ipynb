```python
%sql
SELECT
  product_id,
  product_name,
  category,
  sub_category,
  total_qty_sold,
  total_revenue,
  unique_customers,
  revenue_rank
FROM ecommerce_project_1.ecommerce_project_1.gold_top_products
ORDER BY total_revenue DESC;
```
