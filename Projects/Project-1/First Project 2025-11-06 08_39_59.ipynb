```python
from pyspark.sql.functions import *
from datetime import date,datetime,timedelta

tp=dbutils.widgets.get('Time_to_select')
print(tp)
today=date.today()

if tp=='weekly':
    st=today-timedelta(days=today.weekday(),weeks=1)-timedelta(days=1)
    end=st+timedelta(days=6)
elif tp=='monthly':
    first=today.replace(day=1)
    end=first-timedelta(days=1)
    st=first-timedelta(days=end.day)
print(st,end)    
```
